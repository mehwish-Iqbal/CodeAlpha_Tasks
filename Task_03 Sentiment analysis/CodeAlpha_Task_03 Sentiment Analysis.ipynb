#  Step 5 — Clean Review Text

We clean the review text by converting it to lowercase and removing URLs, punctuation, numbers, and extra spaces. 

This prepares the text for NLP-based sentiment analysis.

### 5.1 Create Text Cleaning Function

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'[^a-z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

### 5.2 Apply Cleaning

```python
df['Clean_Review'] = df['Review Text'].apply(clean_text)
```

### 5.3 Compare Original vs Cleaned Text

```python
df[['Review Text', 'Clean_Review']].head(10)
```

 #  Step 6 — VADER Sentiment Analysis

### 6.1 Import VADER

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()
```

### 6.2 Calculate Sentiment Scores

```python
df['Sentiment_Score'] = df['Clean_Review'].apply(
    lambda text: sia.polarity_scores(text)['compound']
)
```

### 6.3 Create Sentiment Category

```python
def classify_sentiment(score):
    if score >= 0.05:
        return 'Positive'
    elif score <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['Sentiment'] = df['Sentiment_Score'].apply(classify_sentiment)
```

# Step 10 — Emotion Detection

```python
from collections import Counter
import re

# Simple emotion lexicon
emotion_lexicon = {
    'joy': ['love', 'happy', 'great', 'excellent', 'amazing', 'wonderful', 'perfect', 'good'],
    'anger': ['angry', 'hate', 'worst', 'terrible', 'awful', 'frustrated', 'annoying'],
    'sadness': ['sad', 'disappointed', 'unhappy', 'bad', 'poor', 'sorry', 'upset'],
    'fear': ['fear', 'scared', 'afraid', 'worried', 'danger', 'risk', 'unsafe'],
    'surprise': ['surprised', 'unexpected', 'shocked', 'wow', 'suddenly', 'astonished']
}

print("Emotion lexicon created successfully.")
```

```python
def detect_emotions(text):
    words = set(re.findall(r'\b[a-z]+\b', text.lower()))
    
    emotion_counts = {}
    
    for emotion, keywords in emotion_lexicon.items():
        count = sum(1 for word in keywords if word in words)
        emotion_counts[emotion] = count
    
    return emotion_counts


df['Emotion_Scores'] = df['Clean_Review'].apply(detect_emotions)

df[['Clean_Review', 'Emotion_Scores']].head()
```

```python
emotion_totals = Counter()

for scores in df['Emotion_Scores']:
    for emotion, count in scores.items():
        emotion_totals[emotion] += count

emotion_df = pd.DataFrame(
    emotion_totals.items(),
    columns=['Emotion', 'Count']
).sort_values('Count', ascending=False)

emotion_df
```

### Step 10.4 — Emotion Distribution Chart

```python
fig = px.bar(
    emotion_df,
    x='Emotion',
    y='Count',
    text='Count',
    title='Emotion Distribution in Amazon Reviews',
    labels={
        'Emotion': 'Emotion',
        'Count': 'Number of Detected Emotion Words'
    }
)

fig.update_traces(textposition='outside')

fig.update_layout(
    height=600,
    width= 1000,
    title_x=0.5,
    xaxis_title='Emotion',
    yaxis_title='Number of Detected Emotion Words'
)

fig.show()
```

**Insight**

**Joy** is the most frequently detected emotion with **6,553 occurrences**, 
* followed by **sadness (3,291)** and **anger (2,520)**.
* **Surprise** and **fear** appear much less frequently, with **352 and 188 occurrences**,
* respectively. Overall, the results indicate that customer reviews contain a mix of positive and negative emotional expressions, with joy being the most prominent emotion.

# Step 12 — Sentiment Trend Over Time

```python
df['Review Date'] = pd.to_datetime(df['Review Date'], errors='coerce')

df[['Review Date', 'Sentiment']].head()
```

### Step 12.1 — Monthly Sentiment Trend

```python
monthly_sentiment = (
    df.groupby([df['Review Date'].dt.to_period('M'), 'Sentiment'])
      .size()
      .reset_index(name='Count')
)

monthly_sentiment['Review Date'] = monthly_sentiment['Review Date'].astype(str)

monthly_sentiment.head(10)
```

### Step 12.2 — Review Date Range

```python
print("Earliest Review:", df['Review Date'].min())
print("Latest Review:", df['Review Date'].max())
```

### Step 12.3 — 2024 Monthly Sentiment Trend

```python
df_2024 = df[df['Review Date'].dt.year == 2024].copy()

monthly_sentiment_2024 = (
    df_2024.groupby([
        df_2024['Review Date'].dt.to_period('M'),
        'Sentiment'
    ])
    .size()
    .reset_index(name='Count')
)

monthly_sentiment_2024['Review Date'] = (
    monthly_sentiment_2024['Review Date'].astype(str)
)

monthly_sentiment_2024
```

### Step 12.4 — Monthly Sentiment plotly Chart

```python
fig = px.line(
    monthly_sentiment_2024,
    x='Review Date',
    y='Count',
    color='Sentiment',
    markers=True,
    title='Monthly Sentiment Trend in Amazon Reviews - 2024',
    labels={
        'Review Date': 'Month',
        'Count': 'Number of Reviews',
        'Sentiment': 'Sentiment'
    }
)

fig.update_layout(
    height=500,
    width= 1000,
    title_x=0.5,
    xaxis_title='Month',
    yaxis_title='Number of Reviews'
)

fig.show()
```

### **Insight:**

* **Negative:** Increased from **143** reviews in January to a peak of **219** in July.
* **Positive:** Reached its highest level in **July with 161** reviews.
* **Neutral:** Remained relatively low throughout the period, ranging from **16 to 29** reviews.
* **Overall:** Negative sentiment remained higher than positive sentiment in most months of 2024.
